### Value Investing Philosophy

**Value investing** is an investment strategy that involves picking stocks that appear to be trading for less than their intrinsic or book value. The key principles include:

#### 1. Intrinsic Value
Every stock has an underlying "true" value based on its fundamentals (earnings, assets, cash flows). The goal is to estimate this intrinsic value and buy when the market price is below it.

**Formula Concept:**
```
Intrinsic Value = Present Value of Future Cash Flows
```

#### 2. Margin of Safety
To protect against errors in estimation and market volatility, value investors seek a **margin of safety** - typically buying only when the stock trades at 20-30% below estimated intrinsic value.

**Formula:**
```
Margin of Safety = (Intrinsic Value - Market Price) / Intrinsic Value × 100%
```

#### 3. Long-term Focus
Value investing requires patience. The market may take time to recognize a stock's true value. Investors should be prepared to hold positions for 3-5 years or longer.

### CSE-Specific Considerations

- Currency fluctuations (LKR)
- Political and economic factors unique to Sri Lanka
- Sector concentration (banking, manufacturing, plantations)
- Limited analyst coverage compared to developed markets

## 4. Value-Investing Screening

This section filters stocks based on classic value investing criteria. Adjust thresholds based on your risk tolerance and market conditions.

### Screening Criteria:
| Metric | Threshold | Rationale |
|--------|-----------|-----------|
| P/E Ratio | < 12 | Low price relative to earnings |
| P/B Ratio | < 1.2 | Trading near or below book value |
| ROE | > 12% | Efficient use of shareholder equity |
| Dividend Yield | > 4% | Income generation |
| Debt/Equity | < 0.5 | Conservative leverage |
| Free Cash Flow | > 0 | Positive cash generation |

````python
# ============================================
# VALUE INVESTING SCREENING PARAMETERS
# ============================================
# Adjust these thresholds based on your preferences

SCREENING_CRITERIA = {
    'pe_max': 12,           # Maximum P/E ratio
    'pb_max': 1.2,          # Maximum P/B ratio
    'roe_min': 12,          # Minimum ROE (%)
    'dividend_yield_min': 4, # Minimum dividend yield (%)
    'debt_equity_max': 0.5, # Maximum debt/equity ratio
    'fcf_min': 0            # Minimum free cash flow
}

# For capital-intensive sectors (Banking, Utilities), use relaxed debt criteria
CAPITAL_INTENSIVE_SECTORS = ['Banking', 'Utilities', 'Insurance']
CAPITAL_INTENSIVE_DEBT_MAX = 1.0

print("Screening criteria loaded:")
for key, value in SCREENING_CRITERIA.items():
    print(f"  {key}: {value}")
````

````python
def screen_value_stocks(df, criteria, relaxed_sectors=None, relaxed_debt_max=1.0):
    """
    Screen stocks based on value investing criteria.
    
    Parameters:
    -----------
    df : DataFrame
        Stock data with required columns
    criteria : dict
        Dictionary of screening thresholds
    relaxed_sectors : list
        Sectors with relaxed debt criteria
    relaxed_debt_max : float
        Maximum debt/equity for capital-intensive sectors
        
    Returns:
    --------
    DataFrame : Filtered stocks meeting value criteria
    """
    # Create a copy to avoid modifying original
    screened = df.copy()
    
    # Calculate P/E and P/B if not already present
    if 'pe_ratio' not in screened.columns:
        screened['pe_ratio'] = screened['price'] / screened['eps']
    if 'pb_ratio' not in screened.columns:
        screened['pb_ratio'] = screened['price'] / screened['bvps']
    
    # Apply filters
    pe_filter = screened['pe_ratio'] < criteria['pe_max']
    pb_filter = screened['pb_ratio'] < criteria['pb_max']
    roe_filter = screened['roe'] > criteria['roe_min']
    div_filter = screened['dividend_yield'] > criteria['dividend_yield_min']
    fcf_filter = screened['free_cash_flow'] > criteria['fcf_min']
    
    # Debt filter with sector consideration
    if relaxed_sectors and 'sector' in screened.columns:
        standard_debt = (screened['debt_equity'] < criteria['debt_equity_max'])
        relaxed_debt = (screened['sector'].isin(relaxed_sectors)) & (screened['debt_equity'] < relaxed_debt_max)
        debt_filter = standard_debt | relaxed_debt
    else:
        debt_filter = screened['debt_equity'] < criteria['debt_equity_max']
    
    # Combine all filters
    value_mask = pe_filter & pb_filter & roe_filter & div_filter & fcf_filter & debt_filter
    
    # Apply screening
    value_candidates = screened[value_mask].copy()
    
    # Add pass/fail summary for each criterion
    value_candidates['criteria_met'] = 'All'
    
    return value_candidates

# Run the screening
value_candidates = screen_value_stocks(
    df, 
    SCREENING_CRITERIA, 
    relaxed_sectors=CAPITAL_INTENSIVE_SECTORS,
    relaxed_debt_max=CAPITAL_INTENSIVE_DEBT_MAX
)

print(f"\n{'='*60}")
print(f"VALUE CANDIDATES: {len(value_candidates)} stocks found")
print(f"{'='*60}\n")

if len(value_candidates) > 0:
    display_cols = ['ticker', 'company_name', 'sector', 'price', 'pe_ratio', 
                    'pb_ratio', 'roe', 'dividend_yield', 'debt_equity']
    print(value_candidates[display_cols].to_string(index=False))
else:
    print("No stocks meet all value criteria. Consider relaxing thresholds.")
````

## 5. Intrinsic Value Calculations

This section provides three methods to estimate intrinsic value:

1. **Discounted Cash Flow (DCF)** - Based on projected future cash flows
2. **Earnings Power Value (EPV)** - Based on sustainable earnings
3. **Net Asset Value (NAV)** - Based on book value with adjustments

### 5.1 Discounted Cash Flow (DCF) Model

**Formula:**
```
DCF Value = Σ [FCF × (1 + g)^t / (1 + r)^t] + Terminal Value / (1 + r)^n

Where:
- FCF = Free Cash Flow per share
- g = Growth rate
- r = Discount rate (required return)
- n = Projection period (years)
- Terminal Value = FCF_n × (1 + g_terminal) / (r - g_terminal)
```

````python
def calculate_dcf(fcf_per_share, growth_rate=0.05, discount_rate=0.12, 
                  projection_years=5, terminal_growth=0.02):
    """
    Calculate Discounted Cash Flow intrinsic value.
    
    Parameters:
    -----------
    fcf_per_share : float
        Current free cash flow per share
    growth_rate : float
        Expected annual growth rate for projection period (default: 5%)
    discount_rate : float
        Required rate of return / WACC (default: 12%)
    projection_years : int
        Number of years to project (default: 5)
    terminal_growth : float
        Long-term sustainable growth rate (default: 2%)
        
    Returns:
    --------
    float : Estimated intrinsic value per share
    """
    if fcf_per_share <= 0:
        return np.nan  # DCF not applicable for negative cash flows
    
    # Calculate present value of projected cash flows
    pv_fcf = 0
    for year in range(1, projection_years + 1):
        projected_fcf = fcf_per_share * ((1 + growth_rate) ** year)
        pv_fcf += projected_fcf / ((1 + discount_rate) ** year)
    
    # Calculate terminal value
    final_fcf = fcf_per_share * ((1 + growth_rate) ** projection_years)
    terminal_value = final_fcf * (1 + terminal_growth) / (discount_rate - terminal_growth)
    pv_terminal = terminal_value / ((1 + discount_rate) ** projection_years)
    
    # Total intrinsic value
    intrinsic_value = pv_fcf + pv_terminal
    
    return intrinsic_value

# Example calculation
example_fcf = 15.00
example_dcf = calculate_dcf(example_fcf)
print(f"DCF Example:")
print(f"  Free Cash Flow per share: Rs. {example_fcf:.2f}")
print(f"  Estimated Intrinsic Value: Rs. {example_dcf:.2f}")
````

### 5.2 Earnings Power Value (EPV)

**Formula:**
```
EPV = Adjusted Earnings / Cost of Capital

Where:
- Adjusted Earnings = Normalized EPS (excluding one-time items)
- Cost of Capital = Required rate of return
```

EPV assumes no growth and focuses on sustainable earnings power.

````python
def calculate_epv(eps, cost_of_capital=0.12, adjustment_factor=1.0):
    """
    Calculate Earnings Power Value.
    
    Parameters:
    -----------
    eps : float
        Earnings per share
    cost_of_capital : float
        Required rate of return (default: 12%)
    adjustment_factor : float
        Factor to adjust for non-recurring items (default: 1.0)
        
    Returns:
    --------
    float : Estimated intrinsic value per share
    """
    if eps <= 0:
        return np.nan  # EPV not applicable for negative earnings
    
    adjusted_eps = eps * adjustment_factor
    epv = adjusted_eps / cost_of_capital
    
    return epv

# Example calculation
example_eps = 12.50
example_epv = calculate_epv(example_eps)
print(f"EPV Example:")
print(f"  EPS: Rs. {example_eps:.2f}")
print(f"  Estimated Intrinsic Value: Rs. {example_epv:.2f}")
````

### 5.3 Net Asset Value (NAV)

**Formula:**
```
NAV = Book Value per Share × Adjustment Factor

Where:
- Adjustment Factor accounts for hidden assets/liabilities
- Typically ranges from 0.8 to 1.2
```

````python
def calculate_nav(bvps, adjustment_factor=1.0):
    """
    Calculate Net Asset Value.
    
    Parameters:
    -----------
    bvps : float
        Book value per share
    adjustment_factor : float
        Factor to adjust for asset quality (default: 1.0)
        - Use > 1.0 if assets are undervalued (land, investments)
        - Use < 1.0 if assets may be impaired
        
    Returns:
    --------
    float : Estimated intrinsic value per share
    """
    nav = bvps * adjustment_factor
    return nav

# Example calculation
example_bvps = 98.00
example_nav = calculate_nav(example_bvps, adjustment_factor=1.1)
print(f"NAV Example:")
print(f"  Book Value per Share: Rs. {example_bvps:.2f}")
print(f"  NAV (with 10% appreciation): Rs. {example_nav:.2f}")
````

## 6. Valuation Results

This section calculates intrinsic values for all value candidates and compares them to market prices. Stocks with sufficient margin of safety are highlighted.

````python
def calculate_all_valuations(df, dcf_params=None, epv_params=None, nav_params=None):
    """
    Calculate all three valuation methods for each stock.
    
    Parameters:
    -----------
    df : DataFrame
        Stock data with required columns
    dcf_params : dict
        Parameters for DCF calculation
    epv_params : dict
        Parameters for EPV calculation
    nav_params : dict
        Parameters for NAV calculation
        
    Returns:
    --------
    DataFrame : Original data with intrinsic value columns added
    """
    # Default parameters
    if dcf_params is None:
        dcf_params = {'growth_rate': 0.05, 'discount_rate': 0.12, 
                      'projection_years': 5, 'terminal_growth': 0.02}
    if epv_params is None:
        epv_params = {'cost_of_capital': 0.12, 'adjustment_factor': 1.0}
    if nav_params is None:
        nav_params = {'adjustment_factor': 1.0}
    
    result = df.copy()
    
    # Calculate DCF values
    result['dcf_value'] = result['free_cash_flow'].apply(
        lambda x: calculate_dcf(x, **dcf_params)
    )
    
    # Calculate EPV values
    result['epv_value'] = result['eps'].apply(
        lambda x: calculate_epv(x, **epv_params)
    )
    
    # Calculate NAV values
    result['nav_value'] = result['bvps'].apply(
        lambda x: calculate_nav(x, **nav_params)
    )
    
    # Calculate average intrinsic value (excluding NaN)
    result['avg_intrinsic_value'] = result[['dcf_value', 'epv_value', 'nav_value']].mean(axis=1)
    
    # Calculate margin of safety
    result['margin_of_safety'] = ((result['avg_intrinsic_value'] - result['price']) / 
                                   result['avg_intrinsic_value'] * 100)
    
    return result

# Calculate valuations for value candidates
if len(value_candidates) > 0:
    valuations = calculate_all_valuations(value_candidates)
    
    print("\nValuation Results:")
    print("=" * 100)
    
    valuation_cols = ['ticker', 'company_name', 'price', 'dcf_value', 
                      'epv_value', 'nav_value', 'avg_intrinsic_value', 'margin_of_safety']
    print(valuations[valuation_cols].round(2).to_string(index=False))
else:
    print("No value candidates to evaluate.")
    valuations = pd.DataFrame()
````

## 7. Portfolio Construction

This section provides a systematic approach to building a diversified value portfolio based on the recommended allocation:

| Category | Target Allocation | Characteristics |
|----------|------------------|-----------------|
| Stable Dividend Stocks | 40% | High dividend yield, consistent payouts |
| Undervalued Asset-Heavy | 30% | Low P/B, strong balance sheet |
| Export-Oriented | 20% | USD earnings hedge, growth potential |
| Special Situations | 10% | Turnarounds, corporate actions |

````python
# Portfolio allocation targets
PORTFOLIO_ALLOCATION = {
    'stable_dividend': 0.40,      # 40% stable dividend stocks
    'asset_heavy': 0.30,          # 30% undervalued asset-heavy
    'export_oriented': 0.20,      # 20% export-oriented
    'special_situations': 0.10   # 10% special situations
}

# Classification thresholds
CLASSIFICATION_THRESHOLDS = {
    'high_dividend_yield': 4.5,     # >= 4.5% for stable dividend
    'low_pb_threshold': 1.0,        # P/B < 1.0 for asset-heavy
    'export_revenue_threshold': 40  # >= 40% export revenue
}

def classify_stock(row):
    """Classify a stock into portfolio categories."""
    categories = []
    
    # Stable dividend stocks
    if row['dividend_yield'] >= CLASSIFICATION_THRESHOLDS['high_dividend_yield']:
        categories.append('stable_dividend')
    
    # Asset-heavy (undervalued relative to book)
    if row['pb_ratio'] < CLASSIFICATION_THRESHOLDS['low_pb_threshold']:
        categories.append('asset_heavy')
    
    # Export-oriented
    if 'export_revenue_pct' in row and row['export_revenue_pct'] >= CLASSIFICATION_THRESHOLDS['export_revenue_threshold']:
        categories.append('export_oriented')
    
    # If no category, mark as potential special situation
    if not categories:
        categories.append('special_situations')
    
    return categories[0]  # Return primary category

print("Portfolio allocation targets:")
for category, allocation in PORTFOLIO_ALLOCATION.items():
    print(f"  {category}: {allocation*100:.0f}%")
````

````python
def build_portfolio(candidates, total_investment, allocation_targets, max_single_stock=0.15):
    """
    Build a diversified portfolio from value candidates.
    
    Parameters:
    -----------
    candidates : DataFrame
        Screened and valued stocks
    total_investment : float
        Total amount to invest (in LKR)
    allocation_targets : dict
        Target allocation percentages by category
    max_single_stock : float
        Maximum allocation to single stock (default: 15%)
        
    Returns:
    --------
    DataFrame : Portfolio with allocations and share counts
    """
    if len(candidates) == 0:
        print("No candidates available for portfolio construction.")
        return pd.DataFrame()
    
    # Classify stocks
    portfolio = candidates.copy()
    portfolio['category'] = portfolio.apply(classify_stock, axis=1)
    
    # Calculate category budgets
    category_budgets = {cat: total_investment * alloc 
                        for cat, alloc in allocation_targets.items()}
    
    # Initialize allocation columns
    portfolio['target_investment'] = 0.0
    portfolio['shares'] = 0
    portfolio['actual_investment'] = 0.0
    portfolio['portfolio_weight'] = 0.0
    
    # Allocate within each category
    for category in allocation_targets.keys():
        cat_stocks = portfolio[portfolio['category'] == category].copy()
        
        if len(cat_stocks) == 0:
            continue
            
        budget = category_budgets[category]
        
        # Equal weight within category, respecting max single stock limit
        max_per_stock = min(budget / len(cat_stocks), total_investment * max_single_stock)
        per_stock_investment = min(budget / len(cat_stocks), max_per_stock)
        
        for idx in cat_stocks.index:
            price = portfolio.loc[idx, 'price']
            shares = int(per_stock_investment / price)
            actual = shares * price
            
            portfolio.loc[idx, 'target_investment'] = per_stock_investment
            portfolio.loc[idx, 'shares'] = shares
            portfolio.loc[idx, 'actual_investment'] = actual
            portfolio.loc[idx, 'portfolio_weight'] = actual / total_investment * 100
    
    # Remove stocks with zero allocation
    portfolio = portfolio[portfolio['shares'] > 0]
    
    return portfolio

# Build portfolio with Rs. 1,000,000 (1 million LKR)
TOTAL_INVESTMENT = 1_000_000  # Adjust based on your capital

if len(valuations) > 0 and 'margin_of_safety' in valuations.columns:
    # Use all value candidates (or just undervalued if you prefer)
    portfolio_candidates = valuations[valuations['margin_of_safety'] > 0].copy()
    
    portfolio = build_portfolio(
        portfolio_candidates,
        TOTAL_INVESTMENT,
        PORTFOLIO_ALLOCATION
    )
    
    if len(portfolio) > 0:
        print(f"\n{'='*80}")
        print(f"PORTFOLIO CONSTRUCTION (Total Investment: Rs. {TOTAL_INVESTMENT:,.0f})")
        print(f"{'='*80}\n")
        
        display_cols = ['ticker', 'company_name', 'category', 'price', 'shares', 
                        'actual_investment', 'portfolio_weight']
        print(portfolio[display_cols].round(2).to_string(index=False))
        
        print(f"\n{'='*40}")
        print("Portfolio Summary:")
        print(f"{'='*40}")
        print(f"Total Stocks: {len(portfolio)}")
        print(f"Total Invested: Rs. {portfolio['actual_investment'].sum():,.0f}")
        print(f"Cash Remaining: Rs. {TOTAL_INVESTMENT - portfolio['actual_investment'].sum():,.0f}")
        
        print("\nCategory Allocation:")
        category_summary = portfolio.groupby('category')['portfolio_weight'].sum()
        for cat, weight in category_summary.items():
            target = PORTFOLIO_ALLOCATION.get(cat, 0) * 100
            print(f"  {cat}: {weight:.1f}% (target: {target:.0f}%)")
else:
    print("No candidates available for portfolio construction.")
    portfolio = pd.DataFrame()
````

## 8. Risk Management Module

This section implements checks to ensure portfolio risk is managed appropriately.

### Risk Checks:
1. **Concentration Risk**: No single stock > 15% of portfolio
2. **Volatility Warning**: Flag high-volatility stocks
3. **Political Risk**: Warning for sensitive sectors
4. **Auditor Concerns**: Placeholder for qualitative checks

````python
# Risk management parameters
RISK_PARAMETERS = {
    'max_single_stock_pct': 15,          # Maximum 15% in single stock
    'volatility_threshold': 30,           # Flag if annual volatility > 30%
    'high_risk_sectors': ['Gaming', 'Tobacco', 'Alcohol', 'Construction'],
    'political_risk_sectors': ['Utilities', 'Banking', 'Telecom']
}

def check_portfolio_risks(portfolio, risk_params):
    """
    Perform risk checks on the portfolio.
    
    Parameters:
    -----------
    portfolio : DataFrame
        Portfolio with allocations
    risk_params : dict
        Risk parameters and thresholds
        
    Returns:
    --------
    list : List of risk warnings
    """
    warnings_list = []
    
    if len(portfolio) == 0:
        return ["⚠️ No portfolio to analyze"]
    
    # Check 1: Concentration risk
    max_weight = portfolio['portfolio_weight'].max()
    if max_weight > risk_params['max_single_stock_pct']:
        offending = portfolio[portfolio['portfolio_weight'] > risk_params['max_single_stock_pct']]
        for _, row in offending.iterrows():
            warnings_list.append(f"⚠️ CONCENTRATION: {row['ticker']} has {row['portfolio_weight']:.1f}% weight (max: {risk_params['max_single_stock_pct']}%)")
    
    # Check 2: Sector concentration
    if 'sector' in portfolio.columns:
        sector_weights = portfolio.groupby('sector')['portfolio_weight'].sum()
        for sector, weight in sector_weights.items():
            if weight > 40:  # More than 40% in single sector
                warnings_list.append(f"⚠️ SECTOR CONCENTRATION: {sector} has {weight:.1f}% allocation")
    
    # Check 3: Political risk sectors
    if 'sector' in portfolio.columns:
        political_exposure = portfolio[portfolio['sector'].isin(risk_params['political_risk_sectors'])]
        if len(political_exposure) > 0:
            total_political = political_exposure['portfolio_weight'].sum()
            warnings_list.append(f"ℹ️ POLITICAL RISK: {total_political:.1f}% exposed to politically sensitive sectors")
    
    # Check 4: High-risk sectors
    if 'sector' in portfolio.columns:
        high_risk = portfolio[portfolio['sector'].isin(risk_params['high_risk_sectors'])]
        if len(high_risk) > 0:
            for _, row in high_risk.iterrows():
                warnings_list.append(f"⚠️ HIGH-RISK SECTOR: {row['ticker']} is in {row['sector']} sector")
    
    # Check 5: Debt levels
    high_debt = portfolio[portfolio['debt_equity'] > 0.7]
    for _, row in high_debt.iterrows():
        warnings_list.append(f"⚠️ HIGH DEBT: {row['ticker']} has D/E ratio of {row['debt_equity']:.2f}")
    
    if not warnings_list:
        warnings_list.append("✅ No significant risks identified")
    
    return warnings_list

# Run risk checks
if len(portfolio) > 0:
    print("\n" + "="*60)
    print("RISK MANAGEMENT REPORT")
    print("="*60 + "\n")
    
    risk_warnings = check_portfolio_risks(portfolio, RISK_PARAMETERS)
    for warning in risk_warnings:
        print(warning)
else:
    print("No portfolio to analyze for risks.")
````

## 10. Review Framework

This section provides functions to update prices and evaluate whether stocks are still undervalued or if rebalancing is needed.

### Review Frequency:
- **Weekly**: Price updates
- **Monthly**: Full valuation review
- **Quarterly**: Portfolio rebalancing check

````python
def update_prices(portfolio, new_prices):
    """
    Update stock prices in the portfolio.
    
    Parameters:
    -----------
    portfolio : DataFrame
        Current portfolio
    new_prices : dict
        Dictionary of ticker: new_price
        
    Returns:
    --------
    DataFrame : Updated portfolio
    """
    updated = portfolio.copy()
    
    for ticker, new_price in new_prices.items():
        if ticker in updated['ticker'].values:
            mask = updated['ticker'] == ticker
            old_price = updated.loc[mask, 'price'].values[0]
            updated.loc[mask, 'price'] = new_price
            updated.loc[mask, 'price_change_pct'] = ((new_price - old_price) / old_price) * 100
            
            # Recalculate P/E and P/B
            updated.loc[mask, 'pe_ratio'] = new_price / updated.loc[mask, 'eps']
            updated.loc[mask, 'pb_ratio'] = new_price / updated.loc[mask, 'bvps']
    
    return updated

# Example: Update prices (uncomment and modify for actual use)
# new_prices = {
#     'COMB.N0000': 98.00,
#     'JKH.N0000': 190.00,
# }
# portfolio = update_prices(portfolio, new_prices)

print("Price update function ready. Modify the new_prices dict above for updates.")
````

````python
def evaluate_portfolio_status(portfolio, min_margin_of_safety=20):
    """
    Evaluate if stocks are still undervalued and if rebalancing is needed.
    
    Parameters:
    -----------
    portfolio : DataFrame
        Current portfolio with updated prices
    min_margin_of_safety : float
        Minimum acceptable margin of safety
        
    Returns:
    --------
    dict : Status report with recommendations
    """
    report = {
        'still_undervalued': [],
        'fairly_valued': [],
        'overvalued': [],
        'rebalance_needed': False,
        'recommendations': []
    }
    
    if len(portfolio) == 0:
        return report
    
    # Recalculate intrinsic values with current data
    for idx, row in portfolio.iterrows():
        # Simple check based on margin of safety
        if 'margin_of_safety' not in portfolio.columns:
            continue
            
        mos = row.get('margin_of_safety', 0)
        ticker = row['ticker']
        
        if mos >= min_margin_of_safety:
            report['still_undervalued'].append(ticker)
        elif mos >= 0:
            report['fairly_valued'].append(ticker)
            report['recommendations'].append(f"HOLD: {ticker} - margin of safety reduced to {mos:.1f}%")
        else:
            report['overvalued'].append(ticker)
            report['recommendations'].append(f"CONSIDER SELLING: {ticker} - now overvalued by {abs(mos):.1f}%")
            report['rebalance_needed'] = True
    
    # Check category drift
    if 'category' in portfolio.columns:
        category_weights = portfolio.groupby('category')['portfolio_weight'].sum().to_dict()
        for cat, target in PORTFOLIO_ALLOCATION.items():
            actual = category_weights.get(cat, 0) / 100
            if abs(actual - target) > 0.1:  # More than 10% drift
                report['recommendations'].append(
                    f"REBALANCE: {cat} drifted to {actual*100:.0f}% (target: {target*100:.0f}%)"
                )
                report['rebalance_needed'] = True
    
    return report

# Evaluate current portfolio
if len(portfolio) > 0:
    status = evaluate_portfolio_status(portfolio)
    
    print("\n" + "="*60)
    print("PORTFOLIO REVIEW STATUS")
    print("="*60 + "\n")
    
    print(f"Still Undervalued: {len(status['still_undervalued'])} stocks")
    print(f"Fairly Valued: {len(status['fairly_valued'])} stocks")
    print(f"Overvalued: {len(status['overvalued'])} stocks")
    print(f"\nRebalancing Needed: {'Yes' if status['rebalance_needed'] else 'No'}")
    
    if status['recommendations']:
        print("\nRecommendations:")
        for rec in status['recommendations']:
            print(f"  • {rec}")
else:
    print("No portfolio to evaluate.")
````
